Q007 ⭐⭐

Display
emp_name

salary

Salary Band

Rules:

salary >=80000

High

salary >=60000

Medium

otherwise

Low

Use CASE WHEN.

```python
%sql
select e.emp_name, e.salary,
        case 
            when e.salary > 80000 then 'High'
            when e.salary > 60000 then 'Medium'
            else 'Low'
        end as salary_band
    from sql_interview.employees e
        where is_active = 'Y';
```

Q021 ⭐⭐

Find

the top 3 highest salaries.

(No Window Functions)

```python
%sql
-- using simple order by and limit
-- we can't get distinct salary band. will have to use dens rank
select distinct salary
    from sql_interview.employees
    order by salary desc
    limit 3
    
```

```python
%sql
-- using window function dens rank
select distinct salary, rnk from (
    select salary,
        dense_rank() over(order by salary desc) rnk
    from sql_interview.employees
)as employees
where rnk <= 3


```

Q025 ⭐⭐ (Module Review)

Display:

Department

Number of Employees

Average Salary

Highest Salary

Lowest Salary

Only include departments having more than 2 employees

Sort by Average Salary DESC

```python
%sql
select d.dept_name, count(e.dept_id) as number_of_employees, cast(avg(e.salary) as decimal(10,2)) as average_salary, max(e.salary) as highet_salary, min(e.salary) as lowest_salary
    from sql_interview.employees e
        inner join sql_interview.departments d
            on e.dept_id = d.dept_id
    where is_active = 'Y'
    group by d.dept_name
        having count(e.dept_id) > 2
    
    order by avg(e.salary) desc
```
