Q026 ⭐⭐ (Warm-up)

Tables:

employees
departments

Question

Display:

Employee Name
Department Name
Salary

Requirements:

Show only active employees.
Sort by Department Name ascending.
Within each department, sort by Salary DESC.

```python
%sql
select e.emp_name, d.dept_name, e.salary
    from sql_interview.employees e
        inner join departments d
            on e.dept_id = d.dept_id
    
    where e.is_active = 'Y'
    order by d.dept_name, e.salary desc
```

Q029 ⭐⭐

Find departments that currently have no employees.

```python
%sql

-- 1.
select d.dept_name
    from sql_interview.departments d
        left join sql_interview.employees e
            on d.dept_id = e.dept_id
        group by d.dept_name
            having count(e.emp_id) = 0;

-- 2.
select d.dept_name
    from sql_interview.departments d
        left anti join sql_interview.employees e
            on d.dept_id = e.dept_id;

-- 3.
select d.dept_name
    from sql_interview.departments d
        left join sql_interview.employees e
            on d.dept_id = e.dept_id
        where e.dept_id is null;

    
```

Q030 ⭐⭐

Show:

Employee

Manager

using a SELF JOIN.

If an employee has no manager, show NULL.

```python
%sql
select employee.emp_name, manager.emp_name as manager
    from sql_interview.employees employee
        left join sql_interview.employees manager
            on employee.manager_id = manager.emp_id
```

Q031 ⭐⭐⭐

For every manager, show:

Manager Name
Number of Direct Reports

Order by highest number of reports.

```python
%sql

-- EXPLAIN FORMATTED
select manager.emp_id, manager.emp_name as manager, count(employee.emp_id) as direct_reports
    from sql_interview.employees manager
        inner join sql_interview.employees employee
            on manager.emp_id = employee.manager_id
                and manager.emp_id <> employee.emp_id
                and employee.is_active = 'Y'
    -- WHERE manager.emp_id <> employee.emp_id
        group by manager.emp_id, manager.emp_name
            having count(employee.emp_id) > 3
    order by direct_reports desc
    
```

Q032 ⭐⭐⭐

Display every employee along with their department location.

Q033 ⭐⭐⭐

Find employees working in Delhi departments.

(You'll need the department location.)

```python
%sql
select e.emp_name, d.location
    from sql_interview.employees e
        inner join sql_interview.departments d
            on e.dept_id = d.dept_id
                and d.location = 'Delhi'
    where d.location = 'Delhi'
```

Q034 ⭐⭐⭐

Show departments where average employee salary > 60,000.

Return department name instead of department ID.

```python
%sql
select d.dept_name, cast(avg(e.salary) as decimal(10,2)) as avg_salary
    from sql_interview.departments d
        inner join sql_interview.employees e
            on d.dept_id = e.dept_id
                and e.is_active = 'Y'
        group by d.dept_id, d.dept_name
            having avg(e.salary) > 60000
    
```
